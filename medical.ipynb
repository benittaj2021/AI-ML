```python
#!pip install gradio
import gradio as gr
import pandas as pd

# If you already saved new pkl after rename, load it
# If not, just use the scaler and model from memory (run training cells before this)

COLS = ['Age','Gender','HR','SBP','DBP','BS','CKMB','TP'] # after your rename

def predict_heart(Age, Gender, HR, SBP, DBP, BS, CKMB, TP):
    data = [[Age, Gender, HR, SBP, DBP, BS, CKMB, TP]]
    df_input = pd.DataFrame(data, columns=COLS)
    scaled = scaler.transform(df_input)
    pred = model.predict(scaled)[0]
    prob = model.predict_proba(scaled)[0]
    risk = max(prob)*100
    if str(pred).lower() == 'positive' or pred == 1:
        return f"POSITIVE - Heart Attack Risk {risk:.1f}% - Urgent Care Needed"
    else:
        return f"NEGATIVE - Risk {risk:.1f}% - Safe"

# Now Gradio interface
inputs=[
    gr.Slider(minimum=20, maximum=70, value=40, label="Age"),
    gr.Slider(minimum=0, maximum=1, value=1, label="Gender"),
    gr.Slider(minimum=40, maximum=150, value=80, label="HR"),
    gr.Slider(minimum=80, maximum=190, value=120, label="SBP"),
    gr.Slider(minimum=40, maximum=130, value=80, label="DBP"),
    gr.Slider(minimum=50, maximum=300, value=120, label="BS"),
    gr.Slider(minimum=0, maximum=30, value=5, label="CKMB"),
    gr.Slider(minimum=0, maximum=10, value=0.5, label="TP"),
]

interface = gr.Interface(
    fn=predict_heart,
    inputs=inputs,
    outputs=gr.Textbox(label="Prediction"),
    title="Heart attack Prediction App"
)

interface.launch(share=True)
```
